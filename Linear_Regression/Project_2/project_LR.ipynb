```python
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
House_price = pd.read_csv("HousePricePrediction.csv")
```

```python
House_price.head()
```

```python
X= House_price.drop(columns =["Id"])
House_price['SalePrice'] = House_price['SalePrice'].fillna(
  House_price['SalePrice'].mean()) 

# Drop records with null values
House_price = House_price.dropna()

House_price.isnull().sum()
House_price.head()
```

```python
X = House_price.drop(['SalePrice'], axis=1)
y = House_price['SalePrice']
X = pd.get_dummies(X, drop_first=True)

## 2. AAPKA TARIQA (Har column ki apni correct dictionary ke saath):
#X["MSZoning"] = X["MSZoning"].map({"RL": 1, "RM": 0, "C (all)": 2, "FV": 3, "RH": 4})
#X["LotConfig"] = X["LotConfig"].map({"Inside": 0, "Corner": 1, "FR2": 2, "CulDSac": 3, "FR3": 4})
#X["BldgType"] = X["BldgType"].map({"1Fam": 0, "2fmCon": 1, "Duplex": 2, "TwnhsE": 3, "Twnhs": 4})
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2 = r2_score( y_test, y_predict)
print("r-square",  r2)
```
